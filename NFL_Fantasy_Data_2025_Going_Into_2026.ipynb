```python
import pandas as pd

# Define the seasons and weeks we are interested in
last_season = 2025
current_season = 2026
current_weeks = [1, 2]

print(f"Attempting to fetch {last_season} season data (weekly stats) using nflreadpy...")
try:
    # Use load_player_stats to get all weekly data for the last season, using 'seasons' argument
    df_2025_season_weekly_polars = nfl.load_player_stats(seasons=[last_season])
    df_2025_season_weekly = df_2025_season_weekly_polars.to_pandas() # Convert to Pandas DataFrame
    print(f"2025 Season data (weekly stats) for {last_season} loaded successfully!")
except Exception as e:
    print(f"Error fetching {last_season} season data: {e}. Setting to empty DataFrame.")
    df_2025_season_weekly = pd.DataFrame() # An empty DataFrame


print(f"Attempting to fetch {current_season} weekly data for weeks {current_weeks} using nflreadpy...")
try:
    # Load all player stats for the current season and then filter for specific weeks
    df_2026_all_weeks_polars = nfl.load_player_stats(seasons=[current_season])
    df_2026_all_weeks = df_2026_all_weeks_polars.to_pandas() # Convert to Pandas DataFrame
    df_2026_weeks = df_2026_all_weeks[df_2026_all_weeks['week'].isin(current_weeks)]
    print(f"{current_season} Weeks {current_weeks} data loaded successfully!")
except Exception as e:
    print(f"Error fetching {current_season} weekly data for weeks {current_weeks}: {e}. Setting to empty DataFrame.")
    df_2026_weeks = pd.DataFrame() # An empty DataFrame

# Check if dataframes are empty before displaying head or shape
if not df_2025_season_weekly.empty:
    print(f"\n--- Data Loading Summary ---")
    print(f"{last_season} Season Data (Weekly Stats) Head:")
    display(df_2025_season_weekly.head())
    print(f"{last_season} Season Data Shape: {df_2025_season_weekly.shape}")
else:
    print(f"\n--- Data Loading Summary ---")
    print(f"No data loaded for {last_season} season.")

if not df_2026_weeks.empty:
    print(f"\n{current_season} Weeks {current_weeks} Data Head:")
    display(df_2026_weeks.head())
    print(f"{current_season} Weeks {current_weeks} Data Shape: {df_2026_weeks.shape}")
else:
    print(f"No data loaded for {current_season} Weeks {current_weeks}.")
```

## Data Preprocessing and Feature Engineering

Before we can analyze the players, we need to preprocess the data. This involves:
1.  **Understanding the data structure:** Inspecting column names and data types.
2.  **Filtering for relevant positions:** Focusing on offensive players and D/ST.
3.  **Handling missing values:** Addressing any `NaN`s that might affect calculations.
4.  **Feature Engineering:** Calculating standard fantasy football metrics (e.g., total fantasy points, average points per game) and potentially more advanced metrics.
5.  **Combining data:** Merging 2025 and 2026 data for comparison.

```python
# Display columns for 2025 season data
print("\nColumns in 2025 Season Weekly Data:")
print(df_2025_season_weekly.columns.tolist())

# Display columns for 2026 current weeks data
print("\nColumns in 2026 Current Weeks Data:")
print(df_2026_weeks.columns.tolist())

# Display info for both dataframes to check data types and non-null counts
print("\nInfo for 2025 Season Weekly Data:")
df_2025_season_weekly.info()

print("\nInfo for 2026 Current Weeks Data:")
df_2026_weeks.info()
```

### Filter for Relevant Positions and Aggregate Data

We'll define the relevant fantasy positions (QB, RB, WR, TE, and D/ST) and filter our dataframes. Then, we'll aggregate the weekly statistics to get total performance for each player per season (or per selected weeks for the current season). For D/ST, we'll need to identify a specific `position_group` or `position` that represents team defense. Based on the `position_group` values, 'DL', 'LB', 'DB' are defensive players and 'SPEC' is special teams. Since `nflreadpy` provides stats per player, for D/ST we need to consider how to aggregate defensive player stats or if there's a dedicated D/ST entry. Looking at the columns, `nflreadpy` seems to provide player-specific stats. For D/ST, we might need to rely on `fantasy_points` directly associated with a 'DEF' or 'ST' entry if available, or aggregate defensive stats by `team`.

Let's check the unique values in `position` and `position_group` to ensure we capture D/ST correctly.

```python
# Check unique positions and position groups to identify D/ST
print("Unique Positions in 2025 Data:", df_2025_season_weekly['position'].unique())
print("Unique Position Groups in 2025 Data:", df_2025_season_weekly['position_group'].unique())

# Define relevant offensive positions
offensive_positions = ['QB', 'RB', 'WR', 'TE']

# For D/ST, we might need to handle it separately or look for a 'DEF' or 'DST' position.
# Based on the output, 'DEF' is a position, which is good for D/ST.
relevant_positions = offensive_positions + ['DEF']

# Filter 2025 season data for relevant positions
df_2025_offense_defense = df_2025_season_weekly[df_2025_season_weekly['position'].isin(relevant_positions)].copy()

# Filter 2026 current weeks data for relevant positions
df_2026_offense_defense = df_2026_weeks[df_2026_weeks['position'].isin(relevant_positions)].copy()

# Fill NaN fantasy points with 0, as absence of play means 0 points
df_2025_offense_defense['fantasy_points'] = df_2025_offense_defense['fantasy_points'].fillna(0)
df_2025_offense_defense['fantasy_points_ppr'] = df_2025_offense_defense['fantasy_points_ppr'].fillna(0)
df_2026_offense_defense['fantasy_points'] = df_2026_offense_defense['fantasy_points'].fillna(0)
df_2026_offense_defense['fantasy_points_ppr'] = df_2026_offense_defense['fantasy_points_ppr'].fillna(0)

print(f"\nFiltered 2025 data shape: {df_2025_offense_defense.shape}")
print(f"Filtered 2026 data shape: {df_2026_offense_defense.shape}")

# Display a few rows of the filtered data to confirm
print("\n2025 Filtered Data Head:")
display(df_2025_offense_defense.head())
print("\n2026 Filtered Data Head:")
display(df_2026_offense_defense.head())
```

### Aggregate Player Statistics

Now, we'll aggregate the weekly statistics to calculate total fantasy points and average fantasy points per game (PPG) for each player. For D/ST, the aggregation will be by team, as a 'DEF' position usually represents the entire team's defense.

```python
# Aggregation for 2025 Season
# For D/ST, player_name is the team name, so we group by team for DEF positions.
# For offensive players, we group by player_id and player_display_name.

# Define common aggregation columns
common_agg_cols = ['player_id', 'player_name', 'player_display_name', 'position', 'position_group', 'team']

# Aggregate 2025 data
df_2025_agg = df_2025_offense_defense.groupby(common_agg_cols).agg(
    total_games=('week', 'nunique'),
    total_fantasy_points=('fantasy_points', 'sum'),
    total_fantasy_points_ppr=('fantasy_points_ppr', 'sum')
).reset_index()

# Calculate fantasy points per game (PPG)
df_2025_agg['ppg'] = df_2025_agg['total_fantasy_points'] / df_2025_agg['total_games']
df_2025_agg['ppg_ppr'] = df_2025_agg['total_fantasy_points_ppr'] / df_2025_agg['total_games']

# Aggregate 2026 data (current season weeks 1-2)
df_2026_agg = df_2026_offense_defense.groupby(common_agg_cols).agg(
    total_games=('week', 'nunique'),
    total_fantasy_points=('fantasy_points', 'sum'),
    total_fantasy_points_ppr=('fantasy_points_ppr', 'sum')
).reset_index()

# Calculate fantasy points per game (PPG)
df_2026_agg['ppg'] = df_2026_agg['total_fantasy_points'] / df_2026_agg['total_games']
df_2026_agg['ppg_ppr'] = df_2026_agg['total_fantasy_points_ppr'] / df_2026_agg['total_games']

print("\n2025 Aggregated Data Head:")
display(df_2025_agg.head())
print(f"2025 Aggregated Data Shape: {df_2025_agg.shape}")

print("\n2026 Aggregated Data Head:")
display(df_2026_agg.head())
print(f"2026 Aggregated Data Shape: {df_2026_agg.shape}")
```

### Combine and Compare Season Data

To identify players to target, we need to compare their performance from the previous season to the current season. We'll merge the two aggregated dataframes and calculate some comparison metrics, such as the difference in PPG.

```python
# Rename columns for clarity before merging
df_2025_agg_renamed = df_2025_agg.rename(columns={
    'total_games': 'games_2025',
    'total_fantasy_points': 'total_fp_2025',
    'total_fantasy_points_ppr': 'total_fp_ppr_2025',
    'ppg': 'ppg_2025',
    'ppg_ppr': 'ppg_ppr_2025',
    'team': 'team_2025' # Rename team column for 2025
})

df_2026_agg_renamed = df_2026_agg.rename(columns={
    'total_games': 'games_2026',
    'total_fantasy_points': 'total_fp_2026',
    'total_fantasy_points_ppr': 'total_fp_ppr_2026',
    'ppg': 'ppg_2026',
    'ppg_ppr': 'ppg_ppr_2026',
    'team': 'team_2026' # Rename team column for 2026
})

# Define common identifying columns to merge on (excluding team, as it might change per player)
common_id_cols = ['player_id', 'player_name', 'player_display_name', 'position', 'position_group']

# Merge the two dataframes on player identifiers only
# Use an outer join to keep all players from both seasons
players_combined_df = pd.merge(
    df_2025_agg_renamed,
    df_2026_agg_renamed,
    on=common_id_cols,
    how='outer'
)

# Fill NaN values for games played and fantasy points with 0
players_combined_df['games_2025'] = players_combined_df['games_2025'].fillna(0)
players_combined_df['total_fp_2025'] = players_combined_df['total_fp_2025'].fillna(0)
players_combined_df['total_fp_ppr_2025'] = players_combined_df['total_fp_ppr_2025'].fillna(0)
players_combined_df['ppg_2025'] = players_combined_df['ppg_2025'].fillna(0)
players_combined_df['ppg_ppr_2025'] = players_combined_df['ppg_ppr_2025'].fillna(0)

players_combined_df['games_2026'] = players_combined_df['games_2026'].fillna(0)
players_combined_df['total_fp_2026'] = players_combined_df['total_fp_2026'].fillna(0)
players_combined_df['total_fp_ppr_2026'] = players_combined_df['total_fp_ppr_2026'].fillna(0)
players_combined_df['ppg_2026'] = players_combined_df['ppg_2026'].fillna(0)
players_combined_df['ppg_ppr_2026'] = players_combined_df['ppg_ppr_2026'].fillna(0)

# Fill NaN values for team names. 'NoTeam' indicates they didn't play in that season for any team.
players_combined_df['team_2025'] = players_combined_df['team_2025'].fillna('NoTeam')
players_combined_df['team_2026'] = players_combined_df['team_2026'].fillna('NoTeam')

# Calculate the difference in PPG between 2026 and 2025
players_combined_df['ppg_diff'] = players_combined_df['ppg_2026'] - players_combined_df['ppg_2025']
players_combined_df['ppg_ppr_diff'] = players_combined_df['ppg_ppr_2026'] - players_combined_df['ppg_ppr_2025']

# Identify players who changed teams AND played in both seasons
players_combined_df['team_changed'] = False
# Only consider a team change if they played in both seasons and their teams are different
players_combined_df.loc[
    (players_combined_df['games_2025'] > 0) &
    (players_combined_df['games_2026'] > 0) &
    (players_combined_df['team_2025'] != players_combined_df['team_2026']),
    'team_changed'
] = True

print("\nCombined Players Data Head:")
display(players_combined_df.head())
print(f"Combined Players Data Shape: {players_combined_df.shape}")
```

## Player Analysis and Recommendation

With the combined data, we can now identify potential trade targets. We'll focus on players who are performing well currently (2026 PPG) and also consider their performance from last season (2025 PPG) to find players who might be trending up or are undervalued.

### Recommendation Criteria:

For offensive players, we'll look for a balance of:
1.  **Strong Current Performance (2026 PPG)**: Players who are scoring high fantasy points per game in the current season.
2.  **Positive Trend (PPG Difference)**: Players whose current PPG is significantly higher than their last season's PPG, indicating improvement or increased opportunity.
3.  **Last Season's Baseline (2025 PPG)**: Players who also had a decent baseline last season, suggesting a solid foundation.

For D/ST, we'll primarily rank them based on their current season's (2026) fantasy points per game.

```python
# Separate offensive players and D/ST
offensive_players_df = players_combined_df[players_combined_df['position'] != 'DEF'].copy()
dst_df = players_combined_df[players_combined_df['position'] == 'DEF'].copy()

# --- Offensive Player Recommendations ---

# Filter out players who haven't played this season (total_games_2026 > 0)
offensive_players_active_2026 = offensive_players_df[offensive_players_df['games_2026'] > 0].copy()

# Sort by current season's PPG (PPR for now, can be adjusted) and PPG difference
# This prioritizes players with strong current performance and a positive trend
offensive_players_ranked = offensive_players_active_2026.sort_values(
    by=['ppg_ppr_2026', 'ppg_ppr_diff'],
    ascending=[False, False]
).reset_index(drop=True)

# Additionally, filter for players who had at least some activity last season
offensive_players_ranked = offensive_players_ranked[offensive_players_ranked['games_2025'] > 0]


# --- D/ST Recommendations ---
# Sort D/ST by current season's PPG
dst_ranked = dst_df[dst_df['games_2026'] > 0].sort_values(
    by='ppg_2026',
    ascending=False
).reset_index(drop=True)


print("\n--- Top 30 Offensive Players to Target (PPR) ---")
# Display top 30 offensive players, focusing on key columns
display(offensive_players_ranked[['player_display_name', 'position', 'team_2025', 'team_2026', 'ppg_2025', 'ppg_2026', 'ppg_diff', 'ppg_ppr_2025', 'ppg_ppr_2026', 'ppg_ppr_diff']].head(30))

print("\n--- Top 10 D/ST Rankings ---")
# Display top 10 D/ST, focusing on key columns
display(dst_ranked[['player_display_name', 'team_2025', 'team_2026', 'ppg_2025', 'ppg_2026']].head(10))
```

## Categorizing Offensive Player Recommendations

To provide more in-depth insights, let's categorize the offensive players into specific recommendation buckets based on their performance trends:

1.  **Improvement Candidates**: Players showing a significant increase in PPG from last season to this season. These are potential breakout players or those with increased roles.
2.  **Consistent Performers**: Players who maintained a high level of performance from last season and are continuing it into the current season.
3.  **Top Last Year, Underperforming This Year**: Players who had excellent 2025 seasons but have started slow in 2026. These could be 'buy low' candidates if their underlying situation suggests a bounce back.
4.  **Top This Year, New/Breakout**: Players with very high 2026 PPG but either no significant 2025 data or a much lower 2025 PPG. These are true breakout candidates.
5.  **Breakout Candidates (Team Change)**: Players who changed teams between 2025 and 2026 and are showing significant improvement in 2026.

### Note on D/ST Rankings:

It appears the `nflreadpy` library, while excellent for individual player statistics, does not provide aggregated team-level D/ST (Defense/Special Teams) fantasy points in a readily usable format. This means the D/ST ranking section remains empty. If you have another data source for D/ST or a specific methodology for calculating D/ST points from individual defensive player statistics, please let me know, and we can explore that avenue.

```python
# Using the offensive_players_ranked DataFrame from the previous step
# We'll use PPR points for these categorizations for consistency with common fantasy leagues.

# Filter for players who played in both seasons to make comparisons meaningful
comparable_offensive_players = offensive_players_ranked[
    (offensive_players_ranked['games_2025'] > 0) &
    (offensive_players_ranked['games_2026'] > 0)
].copy()


# 1. Improvement Candidates (Significant positive ppg_ppr_diff)
# Let's define 'significant improvement' as ppg_ppr_diff > 3 and current ppg_ppr_2026 > 10
improvement_candidates = comparable_offensive_players[
    (comparable_offensive_players['ppg_ppr_diff'] > 3) &
    (comparable_offensive_players['ppg_ppr_2026'] > 10)
].sort_values(by='ppg_ppr_diff', ascending=False)

improvement_candidates['Rank'] = range(1, len(improvement_candidates) + 1)

print("\n--- Improvement Candidates (PPR) ---")
display(improvement_candidates[['Rank', 'player_display_name', 'position', 'team_2025', 'team_2026', 'ppg_ppr_2025', 'ppg_ppr_2026', 'ppg_ppr_diff']].head(100))


# 2. Consistent Performers (High ppg_ppr_2025 and ppg_ppr_2026, small ppg_ppr_diff)
# High performance > 15 PPR PPG in both seasons, and ppg_ppr_diff between -2 and 2
consistent_performers = comparable_offensive_players[
    (comparable_offensive_players['ppg_ppr_2025'] > 15) &
    (comparable_offensive_players['ppg_ppr_2026'] > 15) &
    (comparable_offensive_players['ppg_ppr_diff'].between(-2, 2))
].sort_values(by='ppg_ppr_2026', ascending=False)

consistent_performers['Rank'] = range(1, len(consistent_performers) + 1)

print("\n--- Consistent Performers (PPR) ---")
display(consistent_performers[['Rank', 'player_display_name', 'position', 'team_2025', 'team_2026', 'ppg_ppr_2025', 'ppg_ppr_2026', 'ppg_ppr_diff']].head(100))


# 3. Top Last Year, Underperforming This Year ('Buy Low' Candidates)
# High ppg_ppr_2025 (> 15) but lower ppg_ppr_2026 (< 10) and significant negative ppg_ppr_diff (< -5)
buy_low_candidates = comparable_offensive_players[
    (comparable_offensive_players['ppg_ppr_2025'] > 15) &
    (comparable_offensive_players['ppg_ppr_2026'] < 10) &
    (comparable_offensive_players['ppg_ppr_diff'] < -5)
].sort_values(by='ppg_ppr_diff', ascending=True)

buy_low_candidates['Rank'] = range(1, len(buy_low_candidates) + 1)

print("\n--- Top Last Year, Underperforming This Year ('Buy Low' Candidates - PPR) ---")
display(buy_low_candidates[['Rank', 'player_display_name', 'position', 'team_2025', 'team_2026', 'ppg_ppr_2025', 'ppg_ppr_2026', 'ppg_ppr_diff']].head(100))


# 4. Top This Year, New/Breakout (High ppg_ppr_2026, little to no 2025 data or large increase)
# For new players, we can check for games_2025 == 0. For breakouts, it's already covered by improvement_candidates.
# Let's focus on players with high ppg_ppr_2026 (> 10) and games_2025 == 0.
# IMPORTANT: Use offensive_players_active_2026 as base, NOT offensive_players_ranked, as ranked filters out games_2025==0
breakout_candidates_new = offensive_players_active_2026[
    (offensive_players_active_2026['games_2025'] == 0) &
    (offensive_players_active_2026['ppg_ppr_2026'] > 10)
].sort_values(by='ppg_ppr_2026', ascending=False)

breakout_candidates_new['Rank'] = range(1, len(breakout_candidates_new) + 1)

print("\n--- Top This Year, New/Breakout Candidates (PPR) ---")
display(breakout_candidates_new[['Rank', 'player_display_name', 'position', 'team_2025', 'team_2026', 'ppg_ppr_2025', 'ppg_ppr_2026', 'ppg_ppr_diff']].head(100))


# 5. Breakout Candidates (Team Change)
# Players who changed teams AND show significant improvement this year
breakout_candidates_team_change = offensive_players_ranked[
    (offensive_players_ranked['team_changed'] == True) &
    (offensive_players_ranked['ppg_ppr_2026'] > 10) &
    (offensive_players_ranked['ppg_ppr_diff'] > 5) # Significant improvement
].sort_values(by='ppg_ppr_diff', ascending=False)

breakout_candidates_team_change['Rank'] = range(1, len(breakout_candidates_team_change) + 1)

print("\n--- Breakout Candidates (Team Change - PPR) ---")
display(breakout_candidates_team_change[['Rank', 'player_display_name', 'position', 'team_2025', 'team_2026', 'ppg_ppr_2025', 'ppg_ppr_2026', 'ppg_ppr_diff']].head(100))
```

```python
print("\n--- Top Last Year, Underperforming This Year ('Buy Low' Candidates - PPR) ---")
display(buy_low_candidates[['Rank', 'player_display_name', 'position', 'team_2025', 'team_2026', 'ppg_ppr_2025', 'ppg_ppr_2026', 'ppg_ppr_diff']].head(100))

print("\n--- Top This Year, New/Breakout Candidates (PPR) ---")
display(breakout_candidates_new[['Rank', 'player_display_name', 'position', 'team_2025', 'team_2026', 'ppg_ppr_2025', 'ppg_ppr_2026', 'ppg_ppr_diff']].head(100))

print("\n--- Breakout Candidates (Team Change - PPR) ---")
display(breakout_candidates_team_change[['Rank', 'player_display_name', 'position', 'team_2025', 'team_2026', 'ppg_ppr_2025', 'ppg_ppr_2026', 'ppg_ppr_diff']].head(100))
```
